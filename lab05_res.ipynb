### 1. Структура абелевих груп

Розглянемо абелеві групи порядку $400$:
$$
\mathbb{Z}_{400}, \quad \mathbb{Z}_{16}\times\mathbb{Z}_{25}, \quad \mathbb{Z}_{20}\times\mathbb{Z}_{20}, \quad \mathbb{Z}_{4}\times\mathbb{Z}_{100}, \quad \mathbb{Z}_{2}\times\mathbb{Z}_{10}\times\mathbb{Z}_{20}, \quad \mathbb{Z}_{505}^{*}.
$$
Для кожної з цих груп виконайте наступні кроки:
1. Знайдіть елементарні дільники групи.
2. Розкладіть групу у добуток циклічних $p$-груп.
3. Визначте кількість елементів кожного можливого порядку.
4. Порівняйте групи та визначте, які з них є ізоморфними.


```sage
groups = [
    CyclicPermutationGroup(400), 
    cartesian_product([CyclicPermutationGroup(16), CyclicPermutationGroup(25)]), 
    cartesian_product([CyclicPermutationGroup(20), CyclicPermutationGroup(20)]), 
    cartesian_product([CyclicPermutationGroup(4), CyclicPermutationGroup(100)]), 
    cartesian_product([CyclicPermutationGroup(2), CyclicPermutationGroup(10), CyclicPermutationGroup(20)]), 
    Integers(505).unit_group()
]
for idx, group in enumerate(groups, 1):
    print(f"Group {idx}: {group}")
    
    try:
        n = group.order()
        print("group order:", n)
        try:
            ab_group = group.as_abelian_group()
            invariants = ab_group.invariants()
            print("Елементарні дільники (структура групи):", invariants)
        except Exception:
            print("Структура задана декартовим добутком.")
        divisors = n.divisors()
        order_counts = {}
        elements = group.list()
        for d in divisors:
            count = 0
            for g in elements:
                try:
                    g_order = g.order()
                except AttributeError:
                    g_order = lcm([x.order() for x in g])
                if g_order == d:
                    count += 1
            if count > 0:
                order_counts[d] = count
        print('orders of the elements:', order_counts)
        print('\n')
    except Exception as e:
        print(f"Error for this group: {e}")
```

### 2. Одна абелева група порядку $n$
1. Реалізуйте функцію, яка для заданого натурального числа $n$ обчислює кількість попарно неізоморфних абелевих груп порядку $n$.
2. Знайдіть перші сто таких натуральних чисел $n$, для яких існує рівно одна абелева група порядку $n$. Для кожного знайденого $n$ проаналізуйте його розклад на прості множники.
3. Сформулюйте гіпотезу: для яких $n\in\mathbb{N}$ існує рівно одна (з точністю до ізоморфізму) абелева група порядку $n$.


```sage
def count_abl_groups(n): 
    factors = factor(n)
    count = 1
    for p, k in factors: 
        count *= number_of_partitions(k)
    return count

unique = []
n = 1
while len(unique) < 100:
    if count_abl_groups(n) == 1:
        unique.append(n)
    n += 1
for i in unique:
    factors = i.factor()
    print(f"n = {i:3d} | factorisation: {factors}")
```

### 4. Факторкільця кільця многочленів
Нехай $f(x)=x^4+x+2$. Виконайте наступні кроки для кожного з кілець $R=\mathbb{Q}, \mathbb{Z}_4, \mathbb{Z}_5$:
1. Побудуйте ідеал $I=(f(x))$ у кільці $R[x]$.
2. Побудуйте факторкільце $R/I$ та знайдіть його порядок.
3. Визначте, чи є факторкільце $R/I$ полем.
4. Знайдіть дільники нуля у факторкільці $R/I$.
5. Розкладіь $f(x)$ на лінійні множники над $R$ та порівняйте з дільниками нуля у факторкільці $R/I$.
6. Перевірте, чи елемент $x+I\in R/I$ є оборотним.

```sage
R_Q.<x> = QQ['x']
f = x^4 + x + 2
I_Q = R_Q.ideal(f)
Q_Q = R_Q.quotient(I_Q)

print("irreducibility of f(x) over Q:", f.is_irreducible())
print("is factroring a field: ", Q_Q.is_field())
print("factorisation f(x):", f.factor())
elem_Q = Q_Q(x)
print("is element x + I a unit:", elem_Q.is_unit())
print("\n")

R_Z4.<x> = Zmod(4)['x']
f_Z4 = x^4 + x + 2
I_Z4 = R_Z4.ideal(f_Z4)
Q_Z4 = R_Z4.quotient(I_Z4)
print("order of a factorring Z4:", Q_Z4.cardinality())
print("factorisation of f(x) over Z4:", f_Z4.factor())
elem_Z4 = Q_Z4(x)
is_unit_Z4 = any(elem_Z4 * y == Q_Z4.one() for y in Q_Z4)
print("is element x + I a unit in Z4:", is_unit_Z4)
print('\n')

R_Z5.<x> = Zmod(5)['x']
f_Z5 = x^4 + x + 2
I_Z5 = R_Z5.ideal(f_Z5)
Q_Z5 = R_Z5.quotient(I_Z5)
print("order of factorring:", Q_Z5.cardinality())
print("is factorring a field:", Q_Z5.is_field())
print("factorisation of f(x) over Z5:", f_Z5.factor())
elem_Z5 = Q_Z5(x)
print("is element x + I a unit in Z5:", elem_Z5.is_unit())
```
