### 1. Властивості циклічних груп

Для наступних порядків $n \in \{100, 173, 530, 746, 911, 1024\}$ розгляньте циклічну групу $G = \mathbb{Z}_n$ (або $C_n$).
1. Знайдіть усі дільники $d$ порядку $n$.<br>
2. Для кожного дільника $d$ обчисліть кількість елементів $a \in G$ таких, що $\text{ord}(a) = d$.<br>
3. Порівняйте отриману кількість з теоретичним значенням $\phi(d)$, де $\phi$ --- функція Ойлера.<br>
4. Знайдіть усі підгрупи групи $G$.<br>
5. Обчисліть загальну кількість усіх підгруп.<br>
6. Порівняйте отриману кількість з теоретичним значенням $\tau(n)$, де $\tau$ -- функція кількості дільників числа $n$.

```sage
p = [100, 173, 530, 746, 911, 1024]
for n in p: 
    Zn = CyclicPermutationGroup(n)
    print("this group is: ", Zn)
    d = divisors(n)
    for k in d: 
        print("number is: ", k)
        number_of_elements = 0
        for a in Zn: 
            if a.order() == k: 
                number_of_elements += 1
        print("number of elements with order ", k, "is: ", number_of_elements)
        print("the euler function of ", k, "is: ", euler_phi(k))
    print("number of all subgroups of Zn is: ", len(Zn.subgroups()))
    print("number of all subgroups using divisors of n: ", len(divisors(n)))
    print("\n")
```

### 2. Порядки елементів в діедральних групах

Для наступних значень $n\in\{5,6,10,12,16,17,20\}$ розгляньте діедральну групу $D_n$.   
1. Знайдіть кількість елементів порядку $d=2$ в групі $D_n$.
2. Для кожного дільника $d>2$ порядку $|D_n|=2n$ знайдіть кількість елементів порядку $d$ у групі $D_n$.<br>

Проаналізуйте результати та дайте відповіді на запитання:
1. Скільки елементів порядку $d=2$ в діедральній групі $D_n$?
2. Скільки елементів порядку $d>2$ в діедральній групі $D_n$?
3. Чи для кожного дільника $d<2n$ порядку групи $D_n$ існує елемент порядку $d$?

```sage
orders = [5, 6, 10, 12, 16, 17, 20]
for n in orders:  
    Dn = DihedralGroup(n)
    number_of_elements = 0
    for a in Dn: 
        if a.order() == 2: 
            number_of_elements += 1
    print("number of elements of order 2 is: ", number_of_elements)
    for div in divisors(2*n): 
        number_of_elements_2 = 0
        if div > 2: 
            for a in Dn: 
                if order(a) == div: 
                    number_of_elements_2 += 1
            print("number of elements of order ", div, "is: ", number_of_elements_2)
    print("\n")
                
```

### 3. Дискретний логарифм

Розглянемо групу $\mathbb{Z}_n^{*}$ для $n=1102158078129785185997827239806$.
1. Знайдіть факторизацію модуля $n$.
2. Перевірте, що група є циклічною, та знайдіть її порядок.
3. Знайдіть найменше натуральне число $a \in \mathbb{N}$ таке, що елемент $\overline{a} \in \mathbb{Z}_n^{*}$ є твірним елементном групи.
4. Перевірте, що порядок елемента $\overline{a}$ дорівнює $\phi(n)$.
5. Виберіть три випадкові елементи $g_1, g_2, g_3 \in \mathbb{Z}_n^{*}$.
6. Для кожного елемента $g \in \{g_1, g_2, g_3\}$ знайдіть дискретний логарифм за модулем $n$, тобто таке число $k$, що $g=\overline{a}^k$ в $\mathbb{Z}_n^{*}$.


```sage
n = 1102158078129785185997827239806
R = Integers(n)
Zn_star= R.unit_group()
generators = R.unit_gens()
print("factorisation of n is: ", factor(n))
if Zn_star.is_cyclic(): 
    print("the group Zn_star is cyclic, with order ", order(Zn_star))
a_conjugate = min(generators)
print("the smallest a conjugate is ", a_conjugate)
if a_conjugate.multiplicative_order() == euler_phi(n): 
    print("order of a conjugate is equal to euler function of n")
g1 = Zn_star.random_element()
g2 = Zn_star.random_element()
g3 = Zn_star.random_element()
elements = [g1, g2, g3]
for idx, g in enumerate(elements, 1):
    elem_ring = R(g) 
    a_ring = R(a_conjugate)
    k = discrete_log(elem_ring, a_ring)
    print(f"element g{idx} = {elem_ring}")
    print(f"discrete log k = {k}")
    is_correct = (a_ring^k == elem_ring)
    print(f"check if (a^k == g) is correct: {is_correct}\n")
```

### 4. Одна лінійна група

Розглянемо підгрупу $G$ групи $GL_3(\mathbb{Z})$, породжену двома матрицями:
$$G = \left\langle \begin{pmatrix} 1 & 0 & 1 \\ 0 & 1 & 0 \\ 0 & 0 & 1 \end{pmatrix}, \begin{pmatrix} 1 & 0 & 0 \\ 0 & 1 & 1 \\ 0 & 0 & 1 \end{pmatrix} \right\rangle.$$

3. Які порядки мають елементи групи $G$?
4. Який вигляд мають елементи групи $G$?
5. Чи є група $G$ циклічною?
6. Якій відомій групі ізоморфна група $G$?
