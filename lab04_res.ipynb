### 1. Класи суміжності та нормальні підгрупи

Розглянемо групу $S_4$ та її підгрупи:<br>
$\quad$ (a) $H_1=\langle (1,2,3,4)\rangle$;<br>
$\quad$ (b) $H_2=\langle (1,3), (1,2,3,4)\rangle$;<br>
$\quad$ (c) $H_3=K_4$.

Для кожної підгрупи $H$ виконайте наступні кроки:
1. Обчисліть порядок підгрупи $H$ та її індекс у групі $[S_4 : H]$.
2. Знайдіть ліві та праві класи суміжності підгрупи $H$ у $S_4$.
3. Визначте, чи є $H$ нормальною підгрупою $S_4$.
4. Якщо $H$ є нормальною, то побудуйте факторгрупу $G/H$, її таблицю Келі та визначіть структуру.
5. Якщо $H$ не є нормальною, то знайдіть такий елемент $a\in S_4$, що $aH\neq Ha$.


```sage
S4 = SymmetricGroup(4)
H1 = S4.subgroup([S4("(1,2,3,4)")])
H2 = S4.subgroup([S4("(1,3)"), S4("(1,2,3,4)")])
H3 = S4.subgroup([S4("(1,2)(3,4)"), S4("(1,3)(2,4)")])
subgroups = [H1, H2, H3]
for H in subgroups:
    is_normal = H.is_normal(S4)
    print(f"is normal in  S4?: {is_normal}")
    if is_normal:
        print("Subgroup is normal. G/H:")
        Q = S4.quotient(H)
        print(f"Order of factor group is: {Q.order()}")
        print(f"Isomorphic? : {Q.structure_description()}")
        print("Ceyley table :")
        print(Q.cayley_table())
    else:
        print("Subgroup is not normal, find aH != Ha:")
        found = False
        for a in S4:
            left_coset = set(a * h for h in H)
            right_coset = set(h * a for h in H)
            if left_coset != right_coset:
                print(f"a = {a}")
                print(f"aH = {list(left_coset)}")
                print(f"Ha = {list(right_coset)}")
                found = True
                print("\n")
                break
```

### 3. Гомоморфізм групи матриць у симетричну групу. 

Розглянемо групу $GL_2(\mathbb{Z}_3)$ та множину прямих $M=\{\ell_1, \ell_2, \ell_3, \ell_4\}$ у просторі $\mathbb{Z}_3^2$, що визначаються парами пропорційних ненульових векторів:
\begin{align*}
\ell_1 = \{ (1,0), (2,0) \}, \
\ell_2 = \{ (0,1), (0,2) \}, \
\ell_3 = \{ (1,1), (2,2) \}, \
\ell_4 = \{ (1,2), (2,1) \}.
\end{align*}

1. Побудуйте групу $GL_2(\mathbb{Z}_3)$.
2. Задайте відображення
    $$
    \phi: GL_2(\mathbb{Z}_3) \longrightarrow S_4,
    $$
    яке переводить матрицю у перестановку, що описує її дію на прямі з $M$.
3. Перевірте, що $\phi$ є гомоморфізмом груп.<br>
4. Обчисліть ядро $N=\ker(\phi)$ та образ $\mathrm{Im}(\phi)$.
5. Побудуйте факторгрупу $GL_2(\mathbb{Z}_3)/N$ та порівняйте її з образом $\mathrm{Im}(\phi)$.



```sage
F = GF(3)
G = GL(2, F)
print(f"order of GL2(Z3): {G.order()}")
l1 = frozenset([(1, 0), (2, 0)])
l2 = frozenset([(0, 1), (0, 2)])
l3 = frozenset([(1, 1), (2, 2)])
l4 = frozenset([(1, 2), (2, 1)])
M = [l1, l2, l3, l4]
S4 = SymmetricGroup(4)
def act_matrix_on_line(A, l):
    transformed_l = set()
    for v in l:
        vec = vector(F, v)
        res = A * vec
        transformed_l.add(tuple(res))
    transformed_l = frozenset(transformed_l)
    for idx, line in enumerate(M):
        if transformed_l == line:
            return idx + 1 
    raise ValueError(f"line {transformed_l} not found in M")
def phi(A):
    images = [act_matrix_on_line(A, l) for l in M]
    return S4(images)
is_homomorphism = True
for A in G:
    for B in G:
        if phi(A * B) != phi(A) * phi(B):
            is_homomorphism = False
            break
print(f"is phi a homomorphism? {is_homomorphism}")
identity_s4 = S4.one()
kernel_elements = [A for A in G if phi(A) == identity_s4]
N = G.subgroup(kernel_elements)
print(f"kernel of N = ker(phi):")
print(f"order of kernel of N: {N.order()}")
print(f"elements of kernel of N:")
for el in N:
    print(el)
image_elements = [phi(A) for A in G]
Im_phi = S4.subgroup(image_elements)
print(f"Im(phi):", Im_phi)
print(f"order Im(phi): {Im_phi.order()}")
print(f"Im(phi) as subgroup of S4: {Im_phi}")
try:
    PG, map_to_PG = G.as_permutation_group()
    N_perm = PG.subgroup([map_to_PG(gen) for gen in N.gens()])
    Q = PG.quo(N_perm)
    print(f"factorgroup GL2(Z3)/N:")
    print(f"order of factorgroup: {Q.order()}")
    is_isomorphic = Q.is_isomorphic(Im_phi)
    print(f"is GL2(Z3)/N isomorphic to Im(phi)? {is_isomorphic}")
except Exception as e:
    print(f"error while buildeing group: {e}")
```
