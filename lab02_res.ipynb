### 3. Підгрупи та їхні властивості

Для кожної з груп $\mathbb{Z}_{24}$, $Q_8$, $D_5$, $S_4$ виконайте наступні кроки:
1. Побудуйте групу.
2. Отримайте та виведіть список усіх підгруп групи.   
3. Визначте кількість підгруп групи з точністю до ізоморфізму.
4. Визначте порядки всіх підгруп групи.
5. Перевірте, чи порядок кожної підгрупи ділить порядок групи.

```sage
Z24 = AdditiveAbelianGroup([24])
G = Z24.permutation_group()
print("group G (order):", G.order())
all_subgroups = G.subgroups()
print(f"\nNumber of subgroups : {len(all_subgroups)}")
isomorphism_classes = {}
for H in all_subgroups:
    iso_type = H.structure_description()
    if iso_type not in isomorphism_classes:
        isomorphism_classes[iso_type] = 0
    isomorphism_classes[iso_type] += 1
print("\nNumber of subgroups with accuracy to isomorphism:")
for iso, count in isomorphism_classes.items():
    print(f"  Isomorphed to {iso}: {count} subgroup(s)")
orders = [H.order() for H in all_subgroups]
print("\norders of all subgroups of Z24:", orders)
group_order = G.order()
divisibility_check = all(group_order % h_order == 0 for h_order in orders)
print(f"\norder Z24: {group_order}")
print("\nLagrange theorem?", divisibility_check)
```

Графом Келі групи $G$ відносно системи твірних $S$ називається орієнтований граф $\Gamma(G,S)$, вершинами якого є елементи групи і кожний $g\in G$ з'єднаний орієнтованим ребро з $gs$ для всіх $s\in S$.
